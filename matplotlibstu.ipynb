```python
import matplotlib.pyplot as plt
import numpy as np
from pylab import mpl
# 设置显示中文字体
mpl.rcParams["font.sans-serif"] = ["SimHei"]
# 构造x轴刻度标签
x_ticks_label = ["11点{}分".format(i) for i in x]
# 构造y轴刻度
y_ticks = range(40)
x = range(0,60)
plt.figure(figsize=(20, 8), dpi=80)
# 模拟y值，温度，0~40之间
y = [np.random.uniform(15,30) for i in x]
# 画折线图
plt.plot(x,y)
y_beijing = [random.uniform(1, 3) for i in x]

# 绘制折线图
plt.plot(x, y_shanghai)
# 使用多次plot可以画多个折线
plt.plot(x, y_beijing, color='r', linestyle='--')

plt.plot(x, y_beijing, label="北京温度")
plt.plot(x, y_shanghai, label="上海温度")
plt.legend(loc=0)


# 修改x,y轴坐标的刻度显示
plt.xticks(x[::5], x_ticks_label[::5])
plt.yticks(y_ticks[::5])
plt.grid(True, linestyle='--', alpha=0.5)
plt.xlabel("时间")
plt.ylabel("温度")
plt.title("中午11点0分到12点之间的温度变化图示", fontsize=20)

# 2.5 添加图例
plt.legend(loc=0)


# 3.图像显示
plt.show()

```

```python
import matplotlib.pyplot as plt
import random
from pylab import mpl
# 中文设置
mpl.rcParams["font.sans-serif"] = ["SimHei"]
mpl.rcParams['axes.unicode_minus'] = False

# 准备数据
x = range(0,60)
y_beijing = [random.uniform(1,3) for i in x]
y_shanghai = [random.uniform(2,4) for i in x]
x_ticks_label = ["11点{}分".format(i) for i in x]

# =========创建画布+2个子坐标系（1行2列）=========
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(20,8), dpi=80)

# =========第一个子图 axes[0] 北京=========
axes[0].plot(x, y_beijing, label="北京温度")
# 设置刻度、标签、标题（面向对象：set_xxx）
axes[0].set_xticks(x[::5])
axes[0].set_xticklabels(x_ticks_label[::5])
axes[0].set_xlabel("时间")
axes[0].set_ylabel("温度")
axes[0].set_title("北京中午11点温度变化")
axes[0].grid(linestyle='--', alpha=0.5)
axes[0].legend(loc=0)

# =========第二个子图 axes[1] 上海=========
axes[1].plot(x, y_shanghai, color="r", label="上海温度")
axes[1].set_xticks(x[::5])
axes[1].set_xticklabels(x_ticks_label[::5])
axes[1].set_xlabel("时间")
axes[1].set_ylabel("温度")
axes[1].set_title("上海中午11点温度变化")
axes[1].grid(linestyle='--', alpha=0.5)
axes[1].legend(loc=0)

plt.show()
```

```python
# 准备刻度数据
x = range(0,60)
x_ticks_label = ["11点{}分".format(i) for i in x]
y_ticks = range(40)

fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(20,8), dpi=80)

# 红框刻度设置
axes[0].set_xticks(x[::5])
axes[0].set_yticks(y_ticks[::5])
axes[0].set_xticklabels(x_ticks_label[::5])

axes[1].set_xticks(x[::5])
axes[1].set_yticks(y_ticks[::5])
axes[1].set_xticklabels(x_ticks_label[::5])

# 网格
axes[0].grid(True, linestyle="--", alpha=0.5)
axes[1].grid(True, linestyle="--", alpha=0.5)

# 坐标轴名称、标题
axes[0].set_xlabel("时间")
axes[0].set_ylabel("温度")
axes[0].set_title("中午11点--12点某城市温度变化图", fontsize=20)
axes[1].set_xlabel("时间")
axes[1].set_ylabel("温度")
axes[1].set_title("中午11点--12点某城市温度变化图", fontsize=20)
axes[0].plot(x, y_beijing, label="北京温度")
axes[1].plot(x, y_shanghai, color="r", label="上海温度")
# # 2.4 图像保存
plt.savefig("./test.png")

# # 2.5 添加图例
# plt.legend(loc=0)
axes[0].legend(loc=0)
axes[1].legend(loc=0)
```

```python
categories = ['A', 'B', 'C', 'D']
values = [3, 7, 5, 4]

plt.bar(categories, values, color='blue')  # 绘制蓝色柱状图
plt.title("Simple Bar Chart")  # 设置图表标题
plt.xlabel("Categories")  # 设置X轴标签
plt.ylabel("Values")  # 设置Y轴标签
plt.show()  # 显示图表
```

```python
# 1. 生成数据
data = np.random.randn(500)  # 生成500个服从标准正态分布的随机数

# 2. 创建画布
plt.figure(figsize=(10, 6))

# 3. 绘制直方图
plt.hist(data, bins=30, color='blue', alpha=0.7, rwidth=0.85)

# 4. 添加标题和标签
plt.title("Histogram of Normally Distributed Data")
plt.xlabel("Value")
plt.ylabel("Frequency")

# 5. 显示图像
plt.grid(True)
plt.show()
```

```python
sizes = [25, 35, 25, 15]
labels = ['Category A', 'Category B', 'Category C', 'Category D']

plt.pie(sizes, labels=labels, autopct='%1.1f%%')  # 绘制饼图，显示百分比
plt.title("Simple Pie Chart")  # 设置图表标题
plt.show()  # 显示图表
```
